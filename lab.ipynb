## ✂️ Step 2: Simulate the data stream from robot controllers.

### 🗣 Instructor Talking Point:
> Simulate the streaming data collection by opening the provided CSV file, and simulate ths stream of robot performance data as if it were coming from controllers, a single reading every 2 seconds.

### 🔧 Your Task:
- For each record read from the CSV file:
    - Open a connection to the database
    - Insert into a table in the database
    - Plot on a chart (refresh it)
    - Close the connection to the database

- Implement the simuation module as an Object-Oriented Python script called `StreamingSimulator`.
- Instantiate an object called `ss` in a Jupyter Notebook.
- Invoke a method called `nextDataPoint` to load a record from the CSV document into a Data Frame for processing.


```python
import os
import pandas as pd
import psycopg2
from dotenv import load_dotenv
from psycopg2.extras import execute_values

load_dotenv()
db_url = os.getenv("DATABASE_URL")
csv_file_path = "./data/RMBR4-2_export_test.csv"
table_name = "rmbr4_export_data"

axis_data = pd.read_csv(csv_file_path)
columns = ["trait"] + [f"axis_{n}" for n in range(1, 15)] + ["time"]
rows = list(axis_data.astype(object).where(axis_data.notna(), None).itertuples(index=False, name=None))

connection = psycopg2.connect(db_url)
try:
    with connection, connection.cursor() as cursor:
        cursor.execute(f"""CREATE TABLE IF NOT EXISTS {table_name} (
            trait VARCHAR(255),
            {", ".join(f"axis_{n} DOUBLE PRECISION" for n in range(1, 15))},
            time TIMESTAMP
        )""")
        # Replace any partial upload so the table matches the CSV exactly.
        cursor.execute(f"TRUNCATE TABLE {table_name}")
        execute_values(
            cursor,
            f"INSERT INTO {table_name} ({', '.join(columns)}) VALUES %s",
            rows,
            page_size=5000,
        )
finally:
    connection.close()
print(f"Uploaded {len(rows):,} rows to Neon table {table_name} in one transaction.")
```

## Stream the Neon table to the dashboard

Replays the full Neon table one reading every 2 seconds into the Plotly dashboard, which is launched in parallel.

```python
import os
import subprocess
import sys
import time
from pathlib import Path

import pandas as pd
import psycopg2
from dotenv import load_dotenv

load_dotenv()
db_url = os.getenv("DATABASE_URL")
table_name = "rmbr4_export_data"
delay_seconds = 2      # one reading per tick, like a robot controller
window_seconds = 90    # amount of history the dashboard receives
start_row = 0          # raise to skip ahead in the table
feed_file = Path("data/live_feed.csv")

# Read the full table from Neon.
connection = psycopg2.connect(db_url)
try:
    full_table = pd.read_sql_query(
        f"SELECT time, {', '.join(f'axis_{n}' for n in range(1, 9))} FROM {table_name} ORDER BY time",
        connection,
    )
finally:
    connection.close()
full_table["time"] = pd.to_datetime(full_table["time"])
print(f"Loaded {len(full_table):,} rows from Neon.")

# Start the dashboard in parallel; it reads the feed file every 2 seconds.
feed_file.unlink(missing_ok=True)
dashboard = subprocess.Popen(
    [sys.executable, "-m", "streamlit", "run", "src/web_ui/web_ui_interface.py", "--server.headless", "true"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL,
)
print("Dashboard: http://localhost:8501 -> choose 'Notebook stream (from Neon)'. Interrupt the kernel to stop.")

try:
    for position in range(start_row, len(full_table)):
        current_time = full_table["time"].iloc[position]
        window = full_table[
            (full_table["time"] <= current_time)
            & (full_table["time"] >= current_time - pd.Timedelta(seconds=window_seconds))
        ]
        temporary_file = feed_file.with_suffix(".tmp")
        window.to_csv(temporary_file, index=False)
        try:
            os.replace(temporary_file, feed_file)
        except PermissionError:
            pass  # the dashboard was reading the file; the next tick will replace it
        print(f"[{time.strftime('%H:%M:%S')}] Streamed row {position + 1}/{len(full_table)}", end="\r")
        time.sleep(delay_seconds)
except KeyboardInterrupt:
    print("\nStream stopped.")
finally:
    dashboard.terminate()
```
