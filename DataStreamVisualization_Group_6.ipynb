```python
import pandas as pd
import numpy as np
import os
import time
import psycopg2
from dotenv import load_dotenv

# Load the credintials of Neon DB connection from the local .env file
load_dotenv()
db_url = os.getenv("DATABASE_URL")
# connect to the Neon postgreSQL database 
connection = psycopg2.connect(db_url)
cursor = connection.cursor()
# Create DB table if it doesn't exist
create_table_query = """CREATE TABLE IF NOT EXISTS rmbr4_export_data (
    trait VARCHAR(255),
    axis_1 DOUBLE PRECISION,
    axis_2 DOUBLE PRECISION,
    axis_3 DOUBLE PRECISION,
    axis_4 DOUBLE PRECISION,
    axis_5 DOUBLE PRECISION,
    axis_6 DOUBLE PRECISION,
    axis_7 DOUBLE PRECISION,
    axis_8 DOUBLE PRECISION,
    axis_9 DOUBLE PRECISION,
    axis_10 DOUBLE PRECISION,
    axis_11 DOUBLE PRECISION,
    axis_12 DOUBLE PRECISION,
    axis_13 DOUBLE PRECISION,
    axis_14 DOUBLE PRECISION,
    time TIMESTAMP
)"""
cursor.execute(create_table_query)

# Migrate an existing table if it was created with an older schema.
# PostgreSQL folds unquoted column names such as Trait to lowercase.
for column, data_type in [
    ("trait", "VARCHAR(255)"),
    ("axis_1", "DOUBLE PRECISION"),
    ("axis_2", "DOUBLE PRECISION"),
    ("axis_3", "DOUBLE PRECISION"),
    ("axis_4", "DOUBLE PRECISION"),
    ("axis_5", "DOUBLE PRECISION"),
    ("axis_6", "DOUBLE PRECISION"),
    ("axis_7", "DOUBLE PRECISION"),
    ("axis_8", "DOUBLE PRECISION"),
    ("axis_9", "DOUBLE PRECISION"),
    ("axis_10", "DOUBLE PRECISION"),
    ("axis_11", "DOUBLE PRECISION"),
    ("axis_12", "DOUBLE PRECISION"),
    ("axis_13", "DOUBLE PRECISION"),
    ("axis_14", "DOUBLE PRECISION"),
    ("time", "TIMESTAMP"),
]:
    cursor.execute(
        f"ALTER TABLE rmbr4_export_data "
        f"ADD COLUMN IF NOT EXISTS {column} {data_type}"
    )

connection.commit()
csv_file_path = './data/RMBR4-2_export_test.csv'

print("CSV file opened to load the data")
axis_data = pd.read_csv(csv_file_path)
print("data stream receiving every 2 seconds")
insert_query = """INSERT INTO rmbr4_export_data (Trait, Axis_1, Axis_2, Axis_3, Axis_4, Axis_5, Axis_6, Axis_7, Axis_8, Axis_9, Axis_10, Axis_11, Axis_12, Axis_13, Axis_14, Time) VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,%s)"""
for index, row in axis_data.iterrows():
    values = tuple(None if pd.isnull(value) else value for value in row)
    cursor.execute(insert_query, values)
    connection.commit()
  #  time.sleep(2)  # Simulate a delay of 2 seconds between each data stream     
print(f"Data uploaded successfully to Neon!")
```
